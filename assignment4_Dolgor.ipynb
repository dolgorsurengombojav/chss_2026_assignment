以下の指示に従って、```Sphere```というクラスを作成しなさい。

- インスタンス化時に```radius```という値を受け取る
- ```surface_area```というメソッドで表面積を計算する

$$S = 4 \pi r^2 $$

- ```volume```というメソッドで体積を計算する


$$V = \frac{4}{3} \pi r^3 $$

Create a class called Sphere following the instructions below.
Receive a value called radius when the object is instantiated.
Create a method called surface_area that calculates the surface area.
Create a method called volume that calculates the volume.


1. Surface Area (表面積)
The formula shown is:
Where:
S = surface area
π (pi) ≈ 3.14159 #constant
r = radius
r square means radius × radius

````python
class Sphere():
    def __init__(self, radius):
        self.radius = radius

    def surface_area(self):
        return 4 * 3.14159 * self.radius ** 2 #means 4 * pi * r^2, which is the formula for the surface area of a sphere.

    def volume(self):
        return (4/3) * 3.14159 * self.radius ** 3   #means (4/3) * pi * r^3, which is the formula for the volume of a sphere.
    
    s = Sphere(1) #meaning the radius of the sphere is 1 unit. 
    print("Surface area of the sphere:", s.surface_area())    
    print("Volume of the sphere:", s.volume())
````
