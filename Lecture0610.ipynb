```python
def divide(dividend, divisor):
    try:
        return dividend / divisor
    except ZeroDivisionError:
        print("Cannot be calculated")

divisors = [10, 5, 0, 20]

for d in divisors:
    result = divide(100, d)
    if result is not None:
        print(result)
```
