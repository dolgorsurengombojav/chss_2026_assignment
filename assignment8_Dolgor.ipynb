## 課題1

以下の図を再現してください

- 年齢の分布を、異なる階級幅を使って、4つのサブプロットをわけて描画する
   - 階級幅を適当で決めてよい
- 各サブプロットには、
    - x軸ラベル、y軸ラベル、サブプロット番号(a-d)を指定する
    - 水平方向のグリッド線を追加する
    - ヒストグラムの色を`lightblue`に変更する

Reproduce figure below
Plot the distribution of age using four separate subplots, each with a different bin width for the histogram.
You may choose the bin widths yourself.
Each subplot should include:
    An x axis label.
    A y axis label.
    A subplot label: (a), (b), (c), and (d).
Add horizontal grid lines to each subplot.
Change the histogram color to light blue.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the Titanic dataset
df = pd.read_csv(
    "https://raw.githubusercontent.com/lvzeyu/css_tohoku/master/"
    "css_tohoku/draft/Data/titanic.csv"
)

# Create a figure with four subplots
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# Different numbers of bins
bins = [5, 10, 20, 30]

# Subplot labels
labels = ["(a)", "(b)", "(c)", "(d)"]

# Draw four histograms
for ax, b, label in zip(axes.flatten(), bins, labels):

    ax.hist(
        df["age"],
        bins=b,
        color="lightblue"
    )

    # Set axis labels
    ax.set_xlabel("Age")
    ax.set_ylabel("Frequency")

    # Add horizontal grid lines
    ax.grid(axis="y")

    # Add subplot label
    ax.set_title(label)

plt.tight_layout()

# Display the figure
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the Titanic dataset
df = pd.read_csv(
    "https://raw.githubusercontent.com/lvzeyu/css_tohoku/master/"
    "css_tohoku/draft/Data/titanic.csv"
)

# Create a figure
plt.figure(figsize=(8, 6))

# Plot the age distribution of all passengers
plt.hist(
    df["age"],
    bins=20,
    color="gray",
    alpha=0.5,
    label="All Passengers"
)

# Plot the age distribution of male passengers
plt.hist(
    df[df["sex"] == "male"]["age"],
    bins=20,
    color="blue",
    alpha=0.6,
    label="Male"
)

# Plot the age distribution of female passengers
plt.hist(
    df[df["sex"] == "female"]["age"],
    bins=20,
    color="red",
    alpha=0.6,
    label="Female"
)

# Add title and axis labels
plt.title("Age Distribution by Sex")
plt.xlabel("Age")
plt.ylabel("Frequency")

# Add a legend
plt.legend()

# Display the figure
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load the Titanic dataset
df = pd.read_csv(
    "https://raw.githubusercontent.com/lvzeyu/css_tohoku/master/"
    "css_tohoku/draft/Data/titanic.csv"
)

# Remove missing values
all_age = df["age"].dropna()
male_age = df[df["sex"] == "male"]["age"].dropna()
female_age = df[df["sex"] == "female"]["age"].dropna()

# Create age bins
bins = np.arange(0, all_age.max() + 5, 5)

# Create the figure
plt.figure(figsize=(8, 6))

# Plot age distribution of all passengers
plt.hist(
    all_age,
    bins=bins,
    color="lightgray",
    edgecolor="black",
    alpha=0.6,
    label="All Passengers"
)

# Plot age distribution of male passengers
plt.hist(
    male_age,
    bins=bins,
    color="steelblue",
    edgecolor="black",
    alpha=0.7,
    label="Male"
)

# Plot age distribution of female passengers
plt.hist(
    female_age,
    bins=bins,
    color="salmon",
    edgecolor="black",
    alpha=0.7,
    label="Female"
)

# Add title and labels
plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Frequency")

# Add legend
plt.legend()

# Display the plot
plt.show()
```
